24) Realizar la operacion 1 << 2 ¿Por qué da ese resultado? ¿Qué es el sistema de numeración binario?

```python
# decimal = 1  binario de 1 = 0001 ( 1 / 2 = cociente 0 residuo 1 entonces el binario de 1 es 1. pero la maquina puede verlo en 4 bit por ejemplo = 0001) 
# """¿Qué es el sistema binario?Es un sistema de numeración en base 2 , donde los únicos dígitos son: 0 y 1"""
1 << 2 #significa que toma el numero 1 se corre 2 bits a la izquierda: 0100. si se realiza la operacion 1 << 2 = 1 * (2 ** 2) = 4
operacion = 1 << 2
print("La operacion de 1 << 2 es: ", operacion)
```
